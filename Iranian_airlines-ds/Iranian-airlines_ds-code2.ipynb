```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('trx-10k.csv')
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head(3)
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df=df[['id','city','amount','card_type','status','time']]
```

```python
text_columns=df.select_dtypes(include='object')
```

```python
for col in text_columns:
    df[col]=df[col].str.strip()
```

```python
for col in text_columns:
    df[col]=df[col].fillna('Unknown')
```

```python
for col in text_columns:
    print(f" {col} : {df[col].isna().sum()}")
```

```python
df['amount']=pd.to_numeric(df['amount'],errors='coerce')
```

```python
df['amount'].min()
```

```python
df=df[df['amount'] > 100]
```

```python
df['amount'].min()
```

```python
df=df[(df['amount'] != 9999999999.0)]
```

```python
df['amount'].max()
```

```python
df['time']=pd.to_datetime(df['time'],errors='coerce')
```

```python
df['city']=df['city'].str.lower()
```

```python
df['city'].unique()
```

```python
update_city={'thran':'tehran',
             'tehr@n':'tehran',
             'thr':'tehran'}
```

```python
df['city']=df['city'].replace(update_city)
```

```python
df['card_type'].unique()
```

```python
df['card_type']=df['card_type'].str.replace(' ','').str.replace('-','').str.lower()
```

```python
update_card={'vsa':'visa',
             'mastcard':'mastercard'}
```

```python
df['card_type']=df['card_type'].replace(update_card)
```

```python
df['status'].unique()
```

```python
update_status={'failed':'fail',
               'succeed':'success'}
df['status']=df['status'].replace(update_status).str.lower()
```

```python
df.info()
```

```python
print('Shape',df.shape)
print('Duplicates',df.duplicated().sum())
print("\nMissing Values")
print(df.isna().sum())
print('\n Dtypes')
print(df.dtypes)
```

```python
df.reset_index(drop=True,inplace=True)
```
