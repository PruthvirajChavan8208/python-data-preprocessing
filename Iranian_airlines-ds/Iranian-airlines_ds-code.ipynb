```python
text_column=df.select_dtypes(include='object')
```

```python
for col in text_column:
    df[col]=df[col].str.strip()
```

```python
for col in text_column:
    df[col]=df[col].fillna('Unknown')
```

```python
df.drop_duplicates()
```

```python
df['status'].unique()
```

```python
change_status={'Success':'success',
               'succeed':'success',
               'FAIL':'fail',
              'failed':'fail'}
```

```python
df['status']=df['status'].replace(change_status)
print(df['status'].unique())
```

```python
df['time']=pd.to_datetime(df['time'],errors='coerce')
```
