```python
ds.isnull().sum() #checking which column have empty or nan values
```

```python
ds=ds.drop(columns=["Unnamed: 2"])
```

```python
ds=ds.drop(columns=["Unnamed: 3","Unnamed: 4"])
```

```python
ds=ds.rename(columns={"v1":"Category","v2":"Message"})
```

```python
ds.duplicated().sum()
```

```python
ds=ds.drop_duplicates(keep="first")
```

```python
ds.shape
```

```python
ds.tail()
```

```python
ds=ds.reset_index(drop=True)
```

```python
mapping={"ham":0,"spam":1}
ds["Category"]=ds["Category"].map(mapping)
```

```python
ds["Message"]=ds["Message"].str.lower()
```

```python
ds["Message"]=ds["Message"].str.split()
```

```python
ds["Message"]=ds["Message"].apply(lambda x:" ".join(x))
ds["Message"]=ds["Message"].str.replace(r"[^a-zA-Z0-9 ]","",regex=True)
ds["Message"]=ds["Message"].str.split()
```
