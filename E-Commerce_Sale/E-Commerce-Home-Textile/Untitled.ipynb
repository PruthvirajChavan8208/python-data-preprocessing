```python
# Errors/Assumptions
# 1.remove 1st and 2nd column because it contain only 39 records
# 2.fill rank sub and rank title with "unranked" and median()/mode()
# 3.price convert to ₹
# 4.fill color count with mean()
# 5.make selling proposition in flaot/int like 1k->1000
# 6.make discount +ve and in flaot
# 7.rename goods-title-link->product title
```

```python
df=df.drop(columns=["goods-title-link--jump","goods-title-link--jump href"])
```

```python
df["rank-title"].isnull().sum()
```

```python
df["rank-title"]=df["rank-title"].fillna("Unranked(New)")
```

```python
df["rank-sub"].isnull().sum()
```

```python
df["rank-sub"]=df["rank-sub"].fillna("General-Home-Textile")
```

```python
df["price"]=df["price"].str[1:]
```

```python
df["price"]=df["price"].fillna("0")
```

```python
df.loc[df["price"]=='888,888.00']
```

```python
df.drop(367,inplace=True)
```

```python
df["price"]=df["price"].astype(float)
```

```python
df["price"]=df["price"]*95
```

```python
df.rename(columns={"price":"Price(₹)","goods-title-link":"Product_Title"})
```

```python
df["color-count"].isnull().sum()
```

```python
df["color-count"].mode()[0]
```

```python
df["color-count"]=df["color-count"].fillna(df["color-count"].mode()[0])
```

```python
df["color-count"].isnull().sum()
```

```python
df["selling_proposition"]=df["selling_proposition"].str[:5]
```

```python
df["selling_proposition"]=df["selling_proposition"].str.replace(r'[+s]',"",regex=True)
```

```python
def convert_k(val):
    str_val=str(val).lower().strip()
    if "k" in str_val:
        return int(float(str_val.replace("k",""))*1000)
    else:
        return str_val if str_val.replace(".","").isdigit() else val
        
```

```python
df["selling_proposition"]=df["selling_proposition"].apply(convert_k)
```

```python
df["selling_proposition"].isnull().sum()
```

```python
df["selling_proposition"]=df["selling_proposition"].fillna(0)
```
