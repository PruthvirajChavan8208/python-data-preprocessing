```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
ds=pd.read_csv("us-shein-jewelry_and_accessories-3548.csv")
```

```python
ds.info()
```

```python
ds.head()
```

```python
ds.tail()
```

```python
# Errors/assumptions
# 1.goods-title link-jump and goods title link href are need to remove
# 2.fill rank title and rank sub with "new"& remove string text 
# 3.price convert to indian ruppe $->₹
# 4.selling proposition must be in numbers
# 5.discount should always in +ve
# 6.rename goods title link-> Product title
```

```python
ds=ds.drop(columns=["goods-title-link--jump","goods-title-link--jump href"])
```

```python
ds["rank-title"].isnull().sum()
```

```python
ds["rank-title"]=ds["rank-title"].fillna("Unranked(New)")
```

```python
ds["rank-sub"].isnull().sum()
```

```python
ds["rank-sub"]=ds["rank-sub"].fillna("General_jewelry")
```

```python
ds["price"]=ds["price"].str[1:]
```

```python
ds["price"]=ds["price"].astype(float)
```

```python
ds["price"]=ds["price"]*95
```

```python
ds["price"].head()
```

```python
ds["selling_proposition"].isnull().sum()
```

```python
ds["selling_proposition"]=ds["selling_proposition"].str[:5]
```

```python
ds["selling_proposition"]=ds["selling_proposition"].str.replace(r"[+s]","",regex=True)
```

```python
def convert_k(val):
    str_val=str(val).lower().strip()
    if "k" in str_val:
        return int(float(str_val.replace("k",""))*1000)
    else:
        return int(float(str_val)) if str_val.replace(".","").isdigit() else val
        
```

```python
ds["selling_proposition"]=ds["selling_proposition"].apply(convert_k)
```

```python
ds["selling_proposition"]=ds["selling_proposition"].fillna(0)
```

```python
ds["discount"]=ds["discount"].str.replace(r"[-%]","",regex=True)
```

```python
ds["discount"]=ds["discount"].astype(float)
```

```python
ds["discount"]=ds["discount"].fillna(0)
```

```python
ds["discount"].unique()
```

```python
ds["goods-title-link"].isnull().sum()
```

```python
ds=ds.dropna(subset=["goods-title-link"])
```

```python
ds.info()
```

```python
ds.duplicated().sum()
```

```python
ds=ds.drop_duplicates()
```

```python
ds=ds.reset_index(drop=True)
```
