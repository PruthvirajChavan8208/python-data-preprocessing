```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("student_performance_updated_1000.csv")
```

```python
df.isnull().sum()
```

```python
df["StudentID"].isna().sum()
```

```python
df=df.dropna(subset=["StudentID"])
```

```python
df=df.dropna(subset=['Name'])
```

```python
df["Gender"].isnull().sum()
```

```python
df["Gender"].mode()[0]
```

```python
df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
df["AttendanceRate"]=df["AttendanceRate"].fillna(df["AttendanceRate"].median())
```

```python
df["StudyHoursPerWeek"]=df["StudyHoursPerWeek"].fillna(df["StudyHoursPerWeek"].median())
```

```python
df["PreviousGrade"].median()
```

```python
df["PreviousGrade"]=df["PreviousGrade"].fillna(df["PreviousGrade"].median())
```

```python
df["ExtracurricularActivities"].median()
```

```python
df["ExtracurricularActivities"]=df["ExtracurricularActivities"].fillna(df["ExtracurricularActivities"].median())
```

```python
df["ParentalSupport"]=df["ParentalSupport"].fillna("Medium")
```

```python
#student who have 20>hrs have median value and who is 20< have min() value
```

```python
g=df[df["FinalGrade"].isna()]
```

```python
print(g[["Name","FinalGrade","StudyHoursPerWeek"]])
```

```python
df["StudyHoursPerWeek"].median()
```

```python
more_hrs_study=df[(df["StudyHoursPerWeek"]>=20) & (df["FinalGrade"].isna())]
```

```python
more_hrs_study[["Name","FinalGrade","StudyHoursPerWeek"]]
```

```python
condition=(df["StudyHoursPerWeek"]>=20) & (df["FinalGrade"].isna())
```

```python
df.loc[condition,"FinalGrade"]=df["FinalGrade"].median()
```

```python
condition_2=(df["StudyHoursPerWeek"]<20) & (df["FinalGrade"].isna())
```

```python
df.loc[condition_2,"FinalGrade"]=df["FinalGrade"].min()
```

```python
df["FinalGrade"].isna().sum()
```

```python
#solution for both study hours & attendance %
# 1.delete
# 2.convert to nan then fill with mean/median
```

```python
df["Study Hours"].isnull().sum()
```

```python
negative_study_hrs=df[df["Study Hours"]<0]
```

```python
print(len(negative_study_hrs))
```

```python
print(negative_study_hrs[['Name',"Study Hours","Attendance (%)"]])
```

```python
df.loc[df["Study Hours"]<0,"Study Hours"]=np.nan
```

```python
df["Study Hours"]=df["Study Hours"].fillna(df["Study Hours"].median())
```

```python
df["Attendance (%)"].isna().sum()
```

```python
high_attendance_rate=df[df["Attendance (%)"]>100]
print(len(high_attendance_rate))
```

```python
print(high_attendance_rate[['Name',"Study Hours","Attendance (%)"]])
```

```python
df.loc[df["Attendance (%)"]>100,"Attendance (%)"]=np.nan
```

```python
df["Attendance (%)"]=df["Attendance (%)"].fillna(df["Attendance (%)"].median())
```

```python
df["Online Classes Taken"].isna().sum()
```

```python
df=df.dropna(subset=["Online Classes Taken"])
```

```python
df.isnull().sum()
```

```python
df=df.reset_index(drop=True)
```

```python
df=df.rename(columns={"AttendanceRate":"Attendance_Rate","StudyHoursPerWeek":"Study_Hours_Per_Week","PreviousGrade":"Previous_Grade"})
```
