```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df=df.drop(columns=['Unnamed: 0','product_title','product_details_dict','mrp','about','technical_details_dict'])
```

```python
df.duplicated().sum()
```

```python
print(f'Size of data set before duplicates delete is:{df.shape}')
```

```python
df=df.drop_duplicates()
```

```python
print(f'size after duplicate delete:{df.shape}')
```

```python
df.shape
```

```python
text_columns=df.select_dtypes(include='object')
```

```python
for col in text_columns:
    df[col]=df[col].str.strip()
```

```python
for col in text_columns:
    df[col]=df[col].fillna('Unknown')
```

```python
num_column=df.select_dtypes(include='number')
```

```python
for col in num_column:
    df[col]=df[col].fillna(0)
```

```python
print(df['rating'].min())
print(df['rating'].max())
```

```python
df.head(2)
```

```python
df['total_rating']=df['total_rating'].str.replace(r'[(),]','',regex=True).str.replace('Unknown','0',regex=False)
```

```python
df['total_rating']=df['total_rating'].astype(int)
```

```python
df['price']=df['price'].str.replace(',','')
```

```python
df['price']=df['price'].astype(float)
```

```python
df['next_month_bought'].unique()
```

```python
df['next_month_bought']=df['next_month_bought'].str.replace('Unknown','0').str.replace(r'\D','',regex=True)
```

```python
df['next_month_bought']=df['next_month_bought'].astype(int)
```

```python
df['offer']=df['offer'].str.replace('Unknown','0').str.replace(r'\D','',regex=True).astype(int)
```

```python
df['brand'].isna().sum()
print(df['brand'].unique())
```

```python
print(df['model_name'].isna().sum())
print(df['model_name'].unique())
```

```python
df['model_name']=df['model_name'].str.split(' ').str[:2].str.join(" ")
```

```python
df['screen_size']=df['screen_size'].str.extract(r'(\d+\d?\d*)')[0]
```

```python
df['screen_size']=df['screen_size'].fillna(0).astype(int)
```

```python
df['colour']=df['colour'].str.split(' ').str[:2].str.join(" ")
```

```python
df['hard_disk_size'].unique()
```

```python
def convert_GB_TB(val):
    str_val=str(val).strip()
    if 'TB' in str_val:
        return int(float(str_val.replace('TB',''))*1000)
    if 'GB' in str_val:
        return str_val.replace('GB','') 
```

```python
df['hard_disk_size']=df['hard_disk_size'].apply(convert_GB_TB)
```

```python
df['hard_disk_size']=df['hard_disk_size'].astype(float)
```

```python
df['cpu_model'].unique()
```

```python
df['ram'].unique()
```

```python
df['ram']=df['ram'].str.replace('Unknown','0').str.replace(r'\D','',regex=True).astype(int)
```
