```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
ds=pd.read_csv("us-shein-appliances-3987.csv")
```

```python
ds.info()
```

```python
ds.head()
```

```python
# #Errors/Assumptions
# 1.goods-title-link--jump & goods-title-link--jump href are having Only 40 values there are too much missing values so we need to delete that columns
# 2. rank-title and rank sub is important in sales data and it is have decent non null valyes so import with values like "unranked" and "general application"
# 3.Price column is in string convert to float also remove symbols like->"$"
# 4.Discount column it is also in string with % remove it & convert to float/int
# 5.selling_proposition columns have 1k+ such text convert 1k with 1000
# 6.goods-title-link it is our product name so rename if you want 
# 7.handle duplicates and missing values in every columns
```

```python
ds=ds.drop(columns=["goods-title-link--jump"])
```

```python
ds=ds.drop(columns=["goods-title-link--jump href"])
```

```python
ds["rank-title"]=ds["rank-title"].fillna("Unranked(New)")
```

```python
ds["rank-title"].isnull().sum()
```

```python
ds["rank-sub"]=ds["rank-sub"].fillna("General Application")
```

```python
ds["price"]=ds["price"].str[1::]
```

```python
ds["price"]=ds["price"].str.replace(",","")
```

```python
ds["price"].unique()
```

```python
ds["price"]=ds["price"].astype(float)
```

```python
ds["price"].dtype
```

```python
ds["discount"].isnull().sum()
```

```python
ds["discount"]=ds["discount"].fillna("0")
```

```python
ds["discount"]=ds["discount"].str.replace("-","")
```

```python
ds["discount"]=ds["discount"].str.replace("%","")
```

```python
ds["discount"]=ds["discount"].astype(int)
```

```python
ds["discount"].unique()
```

```python
ds["discount"].dtype
```

```python
ds["selling_proposition"].isnull().sum()
```

```python
ds["selling_proposition"]=ds["selling_proposition"].fillna("0")
```

```python
ds["selling_proposition"]=ds["selling_proposition"].str[0:4]
```

```python
ds["selling_proposition"]=ds["selling_proposition"].str.replace(r"[+k]","",regex=True)
```

```python
ds["selling_proposition"]=ds["selling_proposition"].astype(float)
```

```python
ds["selling_proposition"].dtype
```

```python
ds["goods-title-link"].isnull().sum()
```

```python
ds=ds.dropna(subset=["goods-title-link"])
```

```python
ds.shape
```

```python
ds=ds.reset_index(drop=True)
```
