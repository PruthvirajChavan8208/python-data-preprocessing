```python
ds.columns=ds.columns.str.strip()
```

```python
ds.duplicated().sum()
```

```python
ds.columns
```

```python
#To remove the extra sppace from text
text_columns=ds.select_dtypes(include='object').columns
```

```python
for col in text_columns:
    ds[col]=ds[col].str.strip()
```

```python
#Missing values checking
# for col in text_columns:
    # print(ds[ds[col].isna()])
```

```python
#missing values handling
for col in text_columns:
    ds[col]=ds[col].fillna('Unknown')
```

```python
ds.isna().sum()
```

```python
ds.select_dtypes(include='number').columns
```

```python
num_columns=ds.select_dtypes(include='number').columns
```

```python
ds[num_columns]=ds[num_columns].fillna(0)
```

```python
#real names of columns
print(type(ds))
print(list(ds.columns))
```

```python
money_columns=[ 'Actual\xa0gross','Adjusted\xa0gross (in 2022 dollars)','Average gross']
```

```python
for col in money_columns:
    ds[col]=(ds[col].str.replace('$','',regex=False).str.replace(',','',regex=False))
    ds[col]=pd.to_numeric(ds[col],errors='coerce')
```

```python
ds['Actual\xa0gross']=ds['Actual\xa0gross'].fillna(0)
```
